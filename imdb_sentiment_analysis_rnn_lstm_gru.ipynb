```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense, LSTM, GRU
```

```python
# Load the IMDb dataset
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)

# Pad sequences to have the same length
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

```python
# Define the RNN model

 # Embedding layer to convert words to vectors
  # RNN layer with 5 units
 # Another RNN layer with 5 units
         # Output layer for binary classification
model = Sequential([
    Input(shape=(100,)),
    Embedding(input_dim=10000, output_dim=32),# 10000 words as input nd we r converting it to 32 vectors
# input= 10000*32+32*5+5*5+5=190+320000 for l1
    #5*5+5*5+5=55
    #5*1+1=6
    SimpleRNN(5, return_sequences=True),#tanh,must be tru except last hidden lyr,kyuki true rkhne se vo next lyr ko o/p pss kr skege

    SimpleRNN(5),#tanh

    Dense(1, activation='sigmoid')
])

model.summary()
```

```python
# Define the LSTM model
model = Sequential([
    Input(shape=(100,)),

    Embedding(input_dim=10000, output_dim=32),

    LSTM(5, return_sequences=True),

    LSTM(5),

    Dense(1, activation='sigmoid')
])

model.summary()
```

```python
# Define the GRU model
model = Sequential([
    Input(shape=(100,)),

    Embedding(input_dim=10000, output_dim=32),

    GRU(5, return_sequences=True),

    GRU(5),

    Dense(1, activation='sigmoid')
])

model.summary()
```

```python
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```
